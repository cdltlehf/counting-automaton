```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from cai4py.instrumentation.constants import OP_NAMES

op_counts_per_run = pd.read_csv('../outputs/op-name-to-counts-polyglot.csv')
merge_sizes = np.load('../outputs/merge-sizes-polyglot.npy')
clone_sizes = np.load('../outputs/clone-sizes-polyglot.npy')
```

```python
mean_counts = op_counts_per_run.mean()
sns.barplot(x=OP_NAMES, y=[mean_counts[op] for op in OP_NAMES], palette="pastel")
plt.savefig('mean-op-counts-polyglot.pdf')
```

```python
max_counts = op_counts_per_run.max()
sns.barplot(x=OP_NAMES, y=[max_counts[op] for op in OP_NAMES], palette="pastel")
plt.savefig('max-op-counts-polyglot.pdf')
```

```python
median_counts = op_counts_per_run.median()
median_counts
```

```python
mean_merge = np.mean(merge_sizes, axis=0)
mean_clone = np.mean(clone_sizes, axis=0)
print("Mean MERGE sizes:", mean_merge)
print("Mean CLONE size:", mean_clone)

median_merge = np.median(merge_sizes, axis=0)
median_clone = np.median(clone_sizes, axis=0)
print("Median MERGE sizes:", median_merge)
print("Median CLONE size:", median_clone)
```

```python
max_merge = max(merge_sizes, key=lambda x: x[0]**2 + x[2]**2)
max_clone = max(clone_sizes, key= lambda x: x[0]**2)
print("Max MERGE size:", max_merge)
print("Max CLONE size:", max_clone)
```
